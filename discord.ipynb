```sql
-- Criação do banco de dados
DROP DATABASE IF EXISTS discord_dos_guri;
CREATE DATABASE discord_dos_guri;
```

```sql
-- Usar o banco de dados criado
USE discord_dos_guri;
```

```sql
-- Tabela de usuários
DROP TABLE IF EXISTS usuarios;
CREATE TABLE usuarios (
    id INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(255) NOT NULL,
    email VARCHAR(255) UNIQUE NOT NULL,
    data_nascimento DATE,
    senha VARCHAR(255) NOT NULL
);
```

```sql
-- Tabela de servidores
DROP TABLE IF EXISTS servidores;
CREATE TABLE servidores (
    id INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(255) NOT NULL,
    criador_id INT NOT NULL,
    FOREIGN KEY (criador_id) REFERENCES usuarios(id)
);

```

```sql
-- Tabela de canais
DROP TABLE IF EXISTS canais;
CREATE TABLE canais (
    id INT PRIMARY KEY AUTO_INCREMENT,
    servidor_id INT NOT NULL,
    nome VARCHAR(255) NOT NULL,
    tipo ENUM('texto', 'voz') NOT NULL,
    FOREIGN KEY (servidor_id) REFERENCES servidores(id)
);
```

```sql
-- Tabela de mensagens
DROP TABLE IF EXISTS mensagens;
CREATE TABLE mensagens (
    id INT PRIMARY KEY AUTO_INCREMENT,
    canal_id INT NOT NULL,
    usuario_id INT NOT NULL,
    conteudo TEXT NOT NULL,
    data_envio DATETIME DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (canal_id) REFERENCES canais(id),
    FOREIGN KEY (usuario_id) REFERENCES usuarios(id)
);
```

```sql
-- Tabela de membros do servidor
DROP TABLE IF EXISTS membros_servidor;
CREATE TABLE membros_servidor (
    usuario_id INT NOT NULL,
    servidor_id INT NOT NULL,
    PRIMARY KEY (usuario_id, servidor_id),
    FOREIGN KEY (usuario_id) REFERENCES usuarios(id),
    FOREIGN KEY (servidor_id) REFERENCES servidores(id)
);
```

```sql
DROP TABLE IF EXISTS amizades;
CREATE TABLE amizades (
    id INT PRIMARY KEY AUTO_INCREMENT,
    usuario1_id INT NOT NULL,
    usuario2_id INT NOT NULL,
    data_criacao TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (usuario1_id) REFERENCES usuarios(id),
    FOREIGN KEY (usuario2_id) REFERENCES usuarios(id),
    CONSTRAINT chk_amizades CHECK (usuario1_id != usuario2_id)
);
```

```sql
-- Inserir 20 amizades aleatórias para cada usuário de forma bidirecional
-- Para o usuário 1
INSERT INTO amizades (usuario1_id, usuario2_id)
SELECT 1 AS usuario1_id, id AS usuario2_id
FROM usuarios
WHERE id != 1
ORDER BY RAND()
LIMIT 20;
```

```sql
SELECT m.id, m.conteudo, m.data_envio, u.nome AS usuario_nome
FROM mensagens m
JOIN usuarios u ON m.usuario_id = u.id
ORDER BY m.data_envio DESC
LIMIT 10 OFFSET 10; -- Segunda página (10 a 20)
```

```sql
SELECT u.nome, COUNT(a.usuario2_id) AS num_seguidores
FROM usuarios u
LEFT JOIN amizades a ON u.id = a.usuario2_id
GROUP BY u.id;
```

```sql
SELECT u.nome
FROM usuarios u
JOIN amizades a ON u.id = a.usuario2_id
GROUP BY u.id
HAVING COUNT(a.usuario2_id) > 2;
```

```sql
CREATE TABLE likes (
    id INT PRIMARY KEY AUTO_INCREMENT,
    mensagem_id INT NOT NULL,
    usuario_id INT NOT NULL,
    data_likada TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (mensagem_id) REFERENCES mensagens(id),
    FOREIGN KEY (usuario_id) REFERENCES usuarios(id)
);
```

```sql
INSERT INTO likes (mensagem_id, usuario_id)
SELECT m.id, u.id
FROM mensagens m
JOIN usuarios u ON u.id != m.usuario_id
ORDER BY RAND()
LIMIT 50;
```

```sql
SELECT m.id, m.conteudo, COUNT(l.id) AS num_likes
FROM mensagens m
LEFT JOIN likes l ON m.id = l.mensagem_id
GROUP BY m.id
ORDER BY num_likes DESC
LIMIT 1;  -- Para pegar a mensagem com o maior número de likes
```

```sql
SELECT m.id, m.conteudo, m.data_envio, u.nome AS usuario_nome
FROM mensagens m
JOIN usuarios u ON m.usuario_id = u.id
ORDER BY m.data_envio DESC
LIMIT 5;
```
