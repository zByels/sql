```sql
SELECT
    COUNT(*) AS total_alunos,
    AVG(idade) AS media_idade,
    AVG(nota) AS media_nota
FROM aluno;
```

```sql
SELECT
    idade,
    COUNT(*) AS n_alunos_idade,
    AVG(nota) AS media_nota
FROM aluno
GROUP BY idade
ORDER BY idade DESC;
```

```sql
SELECT
    idade,
    COUNT(*) AS n_alunos_idade,
    AVG(nota) AS media_nota
FROM aluno
GROUP BY idade
HAVING COUNT(*) > 2
ORDER BY idade DESC;
```

```sql
SELECT
    nome,
    nota
FROM aluno
WHERE nota < (SELECT AVG(nota) FROM aluno);
```

```sql
SELECT
    nome,
    idade,
    nota,
    RANK() OVER (
        PARTITION BY idade 
        ORDER BY nota DESC) AS ranque
FROM aluno
ORDER BY idade, nota DESC;
```

```sql
SELECT
    nome,
    nota,
    AVG(nota) OVER () AS nota_media
FROM aluno
ORDER BY nota DESC;
```

```sql
SELECT
    nome,
    idade,
    nota,
    AVG(nota) OVER (PARTITION BY idade) AS nota_media
FROM aluno
ORDER BY nota DESC;
```

```sql
SELECT
    nome,
    nota,
    (
SELECT AVG(nota)
     FROM aluno
) AS media_nota
FROM aluno
WHERE nota > (
    SELECT AVG(nota)
    FROM aluno
);
```

```sql
WITH aluno_media_geral AS (
SELECT
    nome,
    nota,
    AVG(nota) OVER () AS media_geral
FROM aluno
)
SELECT *
FROM aluno_media_geral
WHERE nota > media_geral;
```

```sql
SELECT
    nome,
    nota,
    CASE
        WHEN nota >= 6 THEN 'Aprovado'
        ELSE 'Reprovado'
    END AS status
FROM aluno;
```
